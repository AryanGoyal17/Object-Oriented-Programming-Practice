## [Concept: Multiple Inheritance and MRO]

```python
class Vehicle:

    def __init__(self, vehicle_number, model, **kwargs):
        super().__init__(**kwargs)
        self.vehicle_number = vehicle_number
        self.model = model

    def vehicle_info(self):

        print(f"Vehicle Number = {self.vehicle_number}")
        print(f"Model = {self.model}")

class ElectricVehicle(Vehicle):

    def __init__(self, battery_capacity, **kwargs):
        super().__init__(**kwargs)
        self.battery_capacity = battery_capacity
    
    def battery_info(self):
        print(f"Battery Capacity = {self.battery_capacity}")

class GPS:

    def __init__(self, latitude, longitude, **kwargs):
        super().__init__(**kwargs)
        self.latitude = latitude
        self.longitude = longitude

    def location_info(self):
        print(f"Latitude = {self.latitude}")
        print(f"Longitude = {self.longitude}")

class SmartCar(ElectricVehicle, GPS):
    def __init__(self, vehicle_number, model, battery_capacity, latitude, longitude):
        super().__init__(vehicle_number=vehicle_number, model=model, battery_capacity = battery_capacity, latitude = latitude, longitude = longitude)
    
    def display_info(self):

        super().vehicle_info()
        super().battery_info()
        super().location_info()

smart_car1 = SmartCar("AB123", "Hyundai", "1000mAH", "92.16 degree", "98.53 degree")

smart_car1.display_info()
print('\n')
print(SmartCar.mro())
```
