# Self driving System (Practice Session - 2, q5)

## Based on diamond inheritance heirarchy

```python
class Vehicle:

    def __init__(self, vin, brand, **kwargs):
        super().__init__(**kwargs)
        self.vin = vin
        self.brand = brand

    def start_engine(self):
        print("Engine has been started")

class Electric(Vehicle):

    def __init__(self, battery_capacity, charge_level, **kwargs):
        super().__init__(**kwargs)
        self.battery_capacity = battery_capacity
        self.charge_level = charge_level
    
    def charge(self):
        print("The Electric Vehicle has been charged")

class Autonomous(Vehicle):

    def __init__(self, software_version, sensor_count, **kwargs):
        super().__init__(**kwargs)
        self.software_version = software_version
        self.sensor_count = sensor_count
    
    def self_drive(self):
        print("Self drive mode is now ON")

class RoboTaxi(Electric, Autonomous):

    def __init__(self, vin, brand, battery_capacity, charge_level, software_version, sensor_count, fare_per_km, current_passenger):
        super().__init__(vin = vin, brand = brand, battery_capacity=battery_capacity, charge_level=charge_level, software_version=software_version, sensor_count=sensor_count)
        self.fare_per_km = fare_per_km
        self.current_passenger = current_passenger
    
    def process_trip(self, distance):
        print(f"Trip started for {self.current_passenger}")
        print(f"Distance = {distance}")
        print(f"Fare per km = {self.fare_per_km}")
        print("Operational checks ---")
        self.start_engine()

    def start_engine(self):
        print('Battery status ---')
        super().charge()
        print('Self driving system status ---')
        super().self_drive()
        print('Engine has been started')

taxi1 = RoboTaxi("Vin1", "Mercedes", "1000mAH", "90%", "v1.2", 2, 100, "Aryan")
taxi2 = RoboTaxi("Vin2", "Audi", "2000mAH", "60%", "v1.1", 1, 120, "Akash")

taxi1.process_trip(15)
taxi2.process_trip(20)
print('\n')
taxi1.start_engine()
taxi2.start_engine()
print('\n')
print(RoboTaxi.__mro__)
```
