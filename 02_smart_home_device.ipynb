## Concepts of Multiple Inheritance and MRO used

```python
class Device:

    def __init__(self, device_name, device_id, **kwargs):
        super().__init__(**kwargs)
        self.device_name = device_name
        self.device_id = device_id

    def device_status(self):
        print("Device is turned ON")

class WifiEnabled:

    def __init__(self, wifi_name, **kwargs):
        super().__init__(**kwargs)
        self.wifi_name = wifi_name
    
    def connect_wifi(self):
        print(f"{self.wifi_name} has been connected!")

class VoiceControlled:

    def __init__(self, assistant_name, **kwargs):
        super().__init__(**kwargs)
        self.assistant_name = assistant_name

    def voice_command(self):
        print(f"{self.assistant_name} is now awake!!")

class SmartSpeaker(Device, WifiEnabled, VoiceControlled):

    def __init__(self, device_name, device_id, wifi_name, assistant_name):
        super().__init__(device_name = device_name, device_id=device_id, wifi_name=wifi_name, assistant_name=assistant_name)
    
    def display_info(self):

        print(f"Device Name = {self.device_name}")
        print(f"Device ID = {self.device_id}")
        print(f"Wifi Name = {self.wifi_name}")
        print(f"Assistant Name = {self.assistant_name}")
        super().device_status()
        super().connect_wifi()
        super().voice_command()


speaker1 = SmartSpeaker("Atlas", "ID - 1", "AryanA35", "Jarvis")

speaker1.display_info()
print('\n')
print(SmartSpeaker.mro())
```
