## Raising a value error during validation in a setter method during encapsulation

### From practice session - 2 [question -> 3]

```python
class Student:

    def __init__(self, name, score):
        self.__name = name
        self.__score = score

    def get_name(self): # name is read-only
        return self.__name
    
    def get_score(self):
        return self.__score

    def set_score(self, new_score):
        if(new_score >= 0 and new_score <= 100):
            self.__score = new_score
            print("Score has been updated!")
        else:
            raise ValueError("Score must be between 0 and 100 [inclusive]")

s1 = Student("Aryan", 94)

print(f"Name = {s1.get_name()}")
print(f"Score = {s1.get_score()}")
s1.set_score(100)
print(f"Score = {s1.get_score()}")
s1.set_score(1000)
print(f"Score = {s1.get_score()}")
```
