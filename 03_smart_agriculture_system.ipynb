## Concepts of Multiple Inheritance used

```python
class Farm:

    def __init__(self, farm_id, location, **kwargs):
        super().__init__(**kwargs)
        self.farm_id = farm_id
        self.location = location
    
    def farm_info(self):
        print(f"Farm ID = {self.farm_id}")
        print(f"Location = {self.location}")

class WeatherSensor:
    
    def __init__(self, temperature, humidity, **kwargs):
        super().__init__(**kwargs)
        self.temperature = temperature
        self.humidity = humidity
    
    def weather_data(self):
        print(f"Temperature = {self.temperature}")
        print(f"Humidity = {self.humidity}")

class SoilSensor:

    def __init__(self, soil_moisture, soil_ph, **kwargs):
        super().__init__(**kwargs)
        self.soil_moisture = soil_moisture
        self.soil_ph = soil_ph
    
    def soil_data(self):
        print(f"Soil Moisture = {self.soil_moisture}")
        print(f"Soil pH = {self.soil_ph}")

class SmartFarm(Farm, WeatherSensor, SoilSensor):
    def __init__(self, farm_id, location, temperature, humidity, soil_moisture, soil_ph):
        super().__init__(farm_id=farm_id, location=location, temperature=temperature, humidity=humidity, soil_moisture=soil_moisture, soil_ph=soil_ph)

    def display(self):

        super().farm_info()
        super().weather_data()
        super().soil_data()

farm1 = SmartFarm("ID -> 1", "Mathura", "30°C", "10 units", "53 units", 7.3)

farm1.display()
print('\n')
print(SmartFarm.mro())
```
