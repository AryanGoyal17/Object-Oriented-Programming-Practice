# University Teaching Assistant Management System [Practice Session - 1 [q5]]

## Concepts used --- Hybrid Inheritance

```python
class Person:

    def __init__(self, name, person_id, **kwargs):
        super().__init__(**kwargs)
        self.name = name
        self.person_id = person_id

class Student(Person):

    def __init__(self, course, gpa, **kwargs):
        super().__init__(**kwargs)
        self.course = course
        self.gpa = gpa

    def calculate_grade(self):
        
        if(self.gpa > 9):
            return 'A'
        elif(self.gpa > 8):
            return 'B'
        else:
            return 'F'


class Employee(Person):

    def __init__(self, department, salary, **kwargs):
        super().__init__(**kwargs)
        self.department = department
        self.salary = salary

    def calculate_pay(self):
        print(f"Total Pay = {self.salary}")

class TeachingAssistant(Student, Employee):

    def __init__(self, name, person_id, course, gpa, department, salary, assigned_subject, hours_worked):
        super().__init__(name=name, person_id=person_id, course=course, gpa=gpa, department=department, salary=salary)
        self.assigned_subject = assigned_subject
        self.hours_worked = hours_worked
    
    def display_ta_profile(self):
        print(f"Name = {self.name}")
        print(f"Id = {self.person_id}")
        print(f"Course = {self.course}")
        print(f"GPA = {self.gpa}")
        print(f"Department = {self.department}")
        print(f"Salary = {self.salary}")
        print(f"Assigned Subject = {self.assigned_subject}")
        print(f"Hours worked = {self.hours_worked}")


ta1 = TeachingAssistant("Aryan", "TA1", "Data Science - I", 9.8, "DSAI", 1000, "Math-I", 5)
ta2 = TeachingAssistant("Akash", "TA2", "Data Science - II", 7.9, "CSE", 980, "Math-II", 3)

print(f"Grade = {ta1.calculate_grade()}")
print('\n')
ta1.calculate_pay()
print('\n')
print(f"Grade = {ta2.calculate_grade()}")
print('\n')
ta2.calculate_pay()
print('\n')
ta1.display_ta_profile()
print('\n')
ta2.display_ta_profile()
print('\n')
print(TeachingAssistant.mro())
```
